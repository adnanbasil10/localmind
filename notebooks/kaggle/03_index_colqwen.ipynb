# 03 — ColQwen2 visual document indexing

SS11 Phase 7. Index on a free T4 (~2 GPU-h for a few thousand pages), store the
vectors, then query on the laptop. No OCR, no chunking, no layout parsing.

**Storage warning:** ~1030 patches x 128 dims per page. Quantize to int8/binary and
report the recall retained — that constraint is itself the engineering story.
